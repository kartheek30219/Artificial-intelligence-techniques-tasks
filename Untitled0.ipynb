```python
from collections import deque

# BFS for A–B–C tree
tree = {
    'A': ['B', 'C'],
    'B': [],
    'C': []
}

def bfs(tree, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for child in tree[node]:
            if child not in visited:
                visited.add(child)
                queue.append(child)

print("BFS Traversal:")
bfs(tree, 'A')
```

```python
# DFS for A–B–C tree
tree = {
    'A': ['B', 'C'],
    'B': [],
    'C': []
}

def dfs(tree, start, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    print(start, end=" ")

    for child in tree[start]:
        if child not in visited:
            dfs(tree, child, visited)

print("DFS Traversal:")
dfs(tree, 'A')
```

```python


heights = [1, 3, 5, 7, 6, 8, 9, 7, 5, 10]
current = 0

print("Starting position:", current)
print("Starting height:", heights[current])

while True:

    left = current - 1
    right = current + 1

    best = current
    if left >= 0 and heights[left] > heights[best]:
        best = left


    if right < len(heights) and heights[right] > heights[best]:
        best = right

    if best == current:
        print("\nNo better neighbor found!")
        print("Hill climbing stopped.")
        break

    current = best

    print("Moved to position:", current,
          "Height:", heights[current])

print("\nFinal position:", current)
print("Final height:", heights[current])
```

```python
from collections import deque

def find_path(rows, cols, start, end):
    queue = deque([start])
    visited = {start}
    parent = {start: None}

    directions = [
        (1, 0),   # down
        (-1, 0),  # up
        (0, 1),   # right
        (0, -1)   # left
    ]

    while queue:
        current = queue.popleft()

        if current == end:
            break

        r, c = current

        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            next_cell = (nr, nc)

            if 0 <= nr < rows and 0 <= nc < cols:
                if next_cell not in visited:
                    visited.add(next_cell)
                    parent[next_cell] = current
                    queue.append(next_cell)

    # Reconstruct path
    if end not in parent:
        return []

    path = []
    current = end

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()
    return path


path = find_path(3, 3, (0, 0), (1, 2))

print("Path:", path)
print("Path length:", len(path) - 1)
```

```python
def find_all_paths(rows, cols, start, end):
    directions = [
        (1, 0),   # down
        (-1, 0),  # up
        (0, 1),   # right
        (0, -1)   # left
    ]

    all_paths = []

    def dfs(current, path, visited):
        if current == end:
            all_paths.append(path.copy())
            return

        r, c = current

        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            next_cell = (nr, nc)

            if (0 <= nr < rows and
                0 <= nc < cols and
                next_cell not in visited):

                visited.add(next_cell)
                path.append(next_cell)

                dfs(next_cell, path, visited)

                # Backtrack
                path.pop()
                visited.remove(next_cell)

    dfs(start, [start], {start})

    return all_paths


# Find all paths
paths = find_all_paths(3, 3, (0, 0), (1, 2))

# Print paths and their lengths
for i, path in enumerate(paths, 1):
    print(f"Path {i}: {path}")
    print(f"Path length: {len(path) - 1}")
    print()
```
